# Sentencia Else (Sino)
Se encadena un if para comprobar el caso contrario (en el que no se cumple la condición)

```python
n = 2
if n%2 ==0:
    print(n, "es un número par")
else:
    print(n, "es unn número impar")
```

## Iteraciones
Iterar significa realizar una acción varias veces. Cada vez que se repite se denomina iteración


## Sentencia While (Mientras)
Se basa en repetir un bloque a partir de evaluar una condición lógica, siempre que ésta sea True.
Queda en manos del programador decidir el momento en que la condición cambie a False para hacer que el While finalice.

```python
 c = 0
 while c <= 5:
    c += 1
    print("c vale: ", c)
```

## Intrucción Break 
Sirve para "romper" la ejecución del While en cualquier momento. No se ejecutará el Else, ya que éste sólo se llama al finalizar la iteración.

```python
c = 0
while c <= 5:
    c += 1
    if (c == 4):
        print ("Rompemos el bucle cuando c vale: ", c)
        break
    print("c vale: ", c)
else:
    print ("Se ha completado toda la iteración y c vale", c)
```

## Intrucción Continue
Sirve para "saltarse" la iteración actual sin romper el bucle.

```python
c = 0
while c <= 5:
    c += 1
    if c == 3 or c == 4: 
        #print ("Rompemos el bucle cuando c vale: ", c)
        continue
    print("c vale: ", c)
else:
    print ("Se ha completado toda la iteración y c vale", c)
```

## Menú Interactivo 

```python
print ("Bienvenidos al menú interactivo")
while (True):
    print ("""¿Qué quieres hacer? Escribe una opción
    1) Saludar
    2) Sumar dos números
    3) Salir""")
    opcion = input ()
    if opcion == '1': 
        print ("Hola, espero que estes bien")
    elif opcion == '2':
        n1 = float (input ("Ingrese el primer valor"))
        n2 = float (input ("Ingrese el segundo valor"))
        print ("El resultado de la sume es: ", n1 + n2)
    elif opcion == '3' :
        print ("¡Hasta luego! Ha sido un placer ayudarte")
    break
else: 
    print ("Comando desconocido, vuelva a intentarlo")
```

1) Realiza un programa que lea dos números por teclado y permita elegir entre 3 opciones en un menú:
- Mostrar una suma de dos números

- Mostrar una multiplicación de los dos números.
- En caso de uno introducir una opción invalida, el programa informará de que no es correcta.

```python
# Ejercicio 1
print ("Menú de opciones")
while (True):
    print ("""Favor seleccione la opción de su agrado
    1) Sumar
    2) Restar
    3) Multiplicar
    4) Salir    
    """)
    eleccion = input()
    if eleccion == '1':
        n1 = float (input ("Favor ingrese el primer número"))
        n2 = float (input ("Favor ingrese el segundo número"))
        print ("La suma de los números es: ", n1+n2)
    elif eleccion == '2':
        n1 = float (input ("Favor ingrese el primer número"))
        n2 = float (input ("Favor ingrese el segundo número"))
        print ("La restra de los números es: ", n1-n2)
    elif eleccion == '3':
        n1 = float (input ("Favor ingrese el primer número"))
        n2 = float (input ("Favor ingrese el segundo número"))
        print ("La multiplicación de los números es: ", n1*n2)
    elif eleccion == '4':
        print ("Gracias por su preferencia (●'◡'●)")
        break
else:
    print ("Opción invalida, favor intentar de nuevo")
```
