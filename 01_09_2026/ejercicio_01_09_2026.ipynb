## Métodos y la palabra self

Self sirve para hacer referencia a los métodos y atributos base de una clase dentro de sus propios métodos.

```python
class Galleta():
    chocolate = False
    def __init__(self):
        print ("Se acaba de crear una galleta")
    
    def chocolatear(self):
        self.chocolate = True

    def tiene_chocolate(self):
        if(self.chocolate):
            print("Soy una galleta chocolateada (●'◡'●)")
        else:
            print("Soy una galleta sin chocolate ಥ_ಥ")
```

```python
g = Galleta()
g.tiene_chocolate()
g.chocolatear()
g.tiene_chocolate
```

# Parámetros en el init (argumentos al instanciar)

```python
class Galleta: 
    chocolate = False

    def __init__(self, sabor, forma):
        self.sabor = sabor
        self.forma = forma
        print("Se acaba de crear una galleta {} y {}".format(sabor,forma))

    def chocolatear(self):
        self.chocolate = True

    def tiene_chocolate(self):
        if(self.chocolate):
            print("Soy una galleta chocolateada")
        else:
            print("Soy una galleta sin chocolate")
```

```python
g = Galleta("salada","cuadrada")
```

## Métodos especiales.

Constructor y destructor

```python
class Pelicula:
    # Constructor de clase (al crear la instancia)
    def __init__(self, titulo, duracion, lanzamiento):
        self.titulo = titulo
        self.duracion = duracion
        self.lanzamiento = lanzamiento
        print("Se ha creado la película", self.titulo)

    #  Destructor  de clase (al borrar la instancia)
    def __del__(self):
        print("Se está borrando la película", self.titulo)
```

```python
p = Pelicula("El padrino",175,1972)
```

```python
# Al reinstanciar la misma variable se crea de nuevo y se borra la anterior

p = Pelicula("El padrino",175,1972)
```

```python
class Pelicula:
    # Constructor de clase (al crear la instancia)
    def __init__(self, titulo, duracion, lanzamiento):
        self.titulo = titulo
        self.duracion = duracion
        self.lanzamiento = lanzamiento
        print("Se ha creado la película", self.titulo)

    #  Destructor  de clase (al borrar la instancia)
    def __del__(self):
        print("Se está borrando la película", self.titulo)

    # Redefinimos el método string
    def __str__(self):
        return "{} lanzada en {} con una duración {} minutos".format(self.titulo,self.lanzamiento,self.duracion)

p = Pelicula("El padrino", 175,1972)
```

1- Crea una clase llamada Persona que tenga los atributos nombre y edad. Agrega un método que muestre sus datos.

```python
class Persona:
    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad
        print("Se ha creado la persona", self.nombre)

    def __del__(self):
        print("Se está borrando a la persona", self.nombre)

    def __str__(self):
        return "{} tiene {} años".format(self.nombre, self.edad)

p = Persona("Carlos", 30)


print(p)
```

2- Crea una clase llamada Rectangulo con los atributos base y altura. Agrega métodos para calcular el área y el perímetro

```python
class Rectangulo:
    def __init__(self, base, altura):
        self.base = base
        self.altura = altura
        print("Se ha creado el rectángulo de {}x{}".format(self.base, self.altura))

    def calcular_area(self):
        return self.base * self.altura

    def calcular_perimetro(self):
        return 2 * (self.base + self.altura)

    def __str__(self):
        return "El rectángulo de base {} y altura {} tiene un área de {} y un perímetro de {}".format(
            self.base, self.altura, self.calcular_area(), self.calcular_perimetro()
        )
```

```python

r = Rectangulo(5, 3)
print(r)
```

3- Crea una clase llamada Estudiante con los atributos nombre, nota1 y nota2. Agrega un método que calcule el promedio e indique si aprobó. La nota mínima es 60.

```python
class Estudiante:
    def __init__(self, nombre, nota1, nota2):
        self.nombre = nombre
        self.nota1 = nota1
        self.nota2 = nota2
        print("Se ha registrado al estudiante", self.nombre)

    def calcular_promedio(self):
        return (self.nota1 + self.nota2) / 2

    def estado_aprobacion(self):
        if self.calcular_promedio() >= 60:
            return "Aprobado"
        else:
            return "Reprobado"

    def __str__(self):
        return "Estudiante: {} | Promedio: {} | Estado: {}".format(
            self.nombre, self.calcular_promedio(), self.estado_aprobacion()
        )
```

```python
e = Estudiante("Ana", 70, 65)
print(e)
```
