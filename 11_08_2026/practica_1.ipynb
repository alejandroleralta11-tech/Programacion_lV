## Listas dentro de listas(listas anidadas)
Podemos manipular fácilmente este tipo de estructuras utilizando múltiples índices, como si nos refieramos a las filas y columnas de una tabla.

```python
a = [1,2,3]
b = [4,5,6]
c = [7,8,9]
r = [a,b,c]
```

```python
r
```

```python
r[0] # Primera sublista
```

```python
r[-1] # Última sublista
```

```python
# De la primera sublista, y de ella, el primer ítem
r[0][0]
```

```python
# De la segunda sublista, y de ella, el segundo elemento
r[1][1]
```

```python
# De la tercera sublista, y de ella, el tercer ítem
r [2][2]
```

```python
# De la última sublista, y de ella, el último ítem
r[-1][-1]
```
